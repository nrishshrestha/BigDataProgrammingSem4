```python
import pandas as pd
import random
from datetime import datetime, timedelta
import math
```

```python
random.seed(42)
```

```python
tt = pd.read_csv('E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\clean_timetabledata.csv')
fares = pd.read_csv('E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\clean_fares.csv')
loc = pd.read_csv('E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\compiled_location_data.csv')
```

```python
lines_timetable = set(tt['LineName'].dropna().unique())
lines_fares = set(fares['LineID'].dropna().str.split(':').str[-1].unique())
lines_location = set(loc['LineRef'].dropna().astype(str).unique())
```

```python
grounded_lines = sorted(lines_timetable & lines_fares & lines_location)
real_stops = tt[['StopPointRef', 'StopName']].dropna().drop_duplicates().values.tolist()
```

```python
print(f"Grounded lines (present in all 3 datasets): {len(grounded_lines)}")
```

```python
SEVERITY_LEVELS = ['verySlight', 'slight', 'normal', 'severe', 'verySevere']
SEVERITY_WEIGHTS = [0.30, 0.30, 0.22, 0.13, 0.05]
```

```python
REASON_TYPES = {
    'RoadworksReason':     ['roadworks', 'resurfacingWork', 'maintenanceWork'],
    'AccidentReason':      ['accident', 'vehicleFire', 'collision'],
    'EquipmentReason':     ['vehicleBreakdown', 'signalFailure'],
    'EnvironmentReason':   ['flooding', 'fog', 'highTemperatures', 'ice'],
    'PersonnelReason':     ['staffSickness', 'staffShortage', 'industrialAction'],
    'MiscellaneousReason': ['diversion', 'specialEvent', 'demonstration'],
}
REASON_CATEGORY_WEIGHTS = [0.35, 0.08, 0.17, 0.10, 0.10, 0.20]
```

```python
N_LINES = len(grounded_lines)
RATE_PER_LINE_PER_DAY = (29.4 * (N_LINES / 135) / 4) / N_LINES
```

```python
START_DATE = datetime(2026, 4, 1)
END_DATE = datetime(2026, 6, 30)
N_DAYS = (END_DATE - START_DATE).days
```

```python
def poisson_sample(lam):
    L = math.exp(-lam)
    k, p = 0, 1.0
    while True:
        k += 1
        p *= random.random()
        if p <= L:
            return k - 1

rows = []
```

```python
rng = random.Random(42)
rows = []
for line_code in grounded_lines:
    n_events = poisson_sample(RATE_PER_LINE_PER_DAY * N_DAYS)
    for _ in range(n_events):
        start_time = START_DATE + timedelta(days=rng.uniform(0, N_DAYS))
        reason_category = rng.choices(list(REASON_TYPES.keys()), weights=REASON_CATEGORY_WEIGHTS, k=1)[0]
        reason_type = rng.choice(REASON_TYPES[reason_category])
        severity = rng.choices(SEVERITY_LEVELS, weights=SEVERITY_WEIGHTS, k=1)[0]
        duration_hours = {
            'verySlight': rng.uniform(0.5, 3),
            'slight':     rng.uniform(1, 8),
            'normal':     rng.uniform(4, 48),
            'severe':     rng.uniform(24, 168),
            'verySevere': rng.uniform(72, 336),
        }[severity]
        end_time = start_time + timedelta(hours=duration_hours)
        planned = reason_category in ('RoadworksReason', 'MiscellaneousReason')
        stop_ref, stop_name = rng.choice(real_stops)     

        rows.append({
            'SituationNumber': f"SYN-{rng.randint(100000,999999)}",   
            'LineRef': line_code,
            'AffectedStopRef': stop_ref,
            'AffectedStopName': stop_name,
            'Severity': severity,
            'ReasonCategory': reason_category,
            'ReasonType': reason_type,
            'Planned': planned,
            'StartTime': start_time.isoformat(),
            'EndTime': end_time.isoformat(),
            'DurationHours': round(duration_hours, 2),
            'IsSynthetic': True,
        })
```

```python
df_disruption = pd.DataFrame(rows)
df_disruption.to_csv('E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\clean_disruptiondata.csv', index=False)
print(f"Generated {len(df_disruption)} disruption records.")
```
