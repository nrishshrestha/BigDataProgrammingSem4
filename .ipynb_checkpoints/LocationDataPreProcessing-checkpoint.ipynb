```python
import pandas as pd
import os
```

```python
data_folder = r"E:\school(softwarica)\Sem-4\BigDataProject\Data\Location"
```

```python
def load_and_compile():
    def get_path(filename):
        return os.path.join(data_folder, filename)

    # loading main dataset
    master_df = pd.read_csv(get_path('all_siri_vm_analysed.csv'))

    before_count = len(master_df)
    master_df = master_df[master_df['OperatorRef'] == 'SCOX'].copy()
    print(f"Filtered to SCOX: {before_count} -> {len(master_df)}rows")

    return master_df
```

```python
df_compiled = load_and_compile()
```

```python
print(f"Success! Compiled {len(df_compiled)} rows.")
print(f"Operators present: {df_compiled['OperatorRef'].unique()}")

output_path = r'E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\compiled_location_data.csv'
os.makedirs(os.path.dirname(output_path), exist_ok=True)
df_compiled.to_csv(output_path, index=False)
print(f"File successfully saved as 'compiled_location_data.csv' in {output_path}")
```
