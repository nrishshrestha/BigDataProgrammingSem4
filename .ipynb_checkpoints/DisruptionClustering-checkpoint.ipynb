```python
disruption = disruption.withColumn(
    "severity_score",
    when(col("Severity") == "verySlight", 1)
    .when(col("Severity") == "slight", 2)
    .when(col("Severity") == "normal", 3)
    .when(col("Severity") == "severe", 4)
    .when(col("Severity") == "verySevere", 5)
).withColumn("week_number", weekofyear(col("StartTime")))
```

```python
disruption.createOrReplaceTempView("disruption")
```

```python
query_features = """
SELECT
    LineRef AS line_code,
    week_number,
    COUNT(*) AS disruption_count,
    AVG(severity_score) AS avg_severity,
    AVG(DurationHours) AS avg_duration_hours,
    AVG(CASE WHEN Planned THEN 1 ELSE 0 END) AS pct_planned
FROM disruption
GROUP BY LineRef, week_number
"""
```

```python
lines_df = timetable.select("LineName").distinct().withColumnRenamed("LineName", "line_code")
```

```python
disruption_features = spark.sql(query_features)
```

```python
START_DATE = datetime(2026, 4, 1)
END_DATE = datetime(2026, 6, 30)
week_numbers = sorted(set(
    (START_DATE + timedelta(days=d)).isocalendar()[1]
    for d in range((END_DATE - START_DATE).days + 1)
))
```

```python
weeks_df = spark.createDataFrame([(w,) for w in week_numbers], ["week_number"])
full_grid = lines_df.crossJoin(weeks_df)
print("Full grid size:", full_grid.count())
```

```python
feature_table = full_grid.join(
    disruption_features, on=["line_code", "week_number"], how="left"
).fillna({
    "disruption_count": 0, "avg_severity": 0.0,
    "avg_duration_hours": 0.0, "pct_planned": 0.0
})
```

```python
feature_table = feature_table.withColumn(
    "log_avg_duration_hours", log1p(col("avg_duration_hours"))
)
```

```python
df = spark.read.csv(r"E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\clustering_features.csv", header=True, inferSchema=True)
print("Before filter:", df.count())
```

```python
df = df.filter(col("disruption_count") > 0)
print("After filter (real signal only):", df.count())
```

```python
FEATURE_COLS = ["disruption_count", "avg_severity", "log_avg_duration_hours", "pct_planned"]
assembler_raw = VectorAssembler(inputCols=FEATURE_COLS, outputCol="features_raw")
df_raw = assembler_raw.transform(df)
```

```python
def rank_clusters_low_to_high(df_pd, cluster_col, metric_col="disruption_count"):
    order = (df_pd.groupby(cluster_col)[metric_col].mean()
             .sort_values().index.tolist())
    names = ["Low Disruptions", "Moderate Disruptions", "High Disruptions"]
    if len(order) != 3:
        raise ValueError(f"Expected 3 clusters, got {len(order)}: {order}")
    return dict(zip(order, names))
```

```python
#KMEANS
kmeans_baseline = KMeans(featuresCol="features_raw", predictionCol="cluster_baseline", k=3, seed=42)
model_baseline = kmeans_baseline.fit(df_raw)
pred_baseline = model_baseline.transform(df_raw)
wcss_baseline = model_baseline.summary.trainingCost

scaler = StandardScaler(inputCol="features_raw", outputCol="features_scaled", withMean=True, withStd=True)
scaler_model = scaler.fit(df_raw)
df_scaled = scaler_model.transform(df_raw)
pred_baseline_scaled = scaler_model.transform(pred_baseline)

evaluator_fair = ClusteringEvaluator(featuresCol="features_scaled", predictionCol="cluster_baseline", metricName="silhouette")
silhouette_baseline = evaluator_fair.evaluate(pred_baseline_scaled)

pred_baseline_pd = pred_baseline.select("line_code", "week_number", "disruption_count", "cluster_baseline").toPandas()
cluster_map_kmeans = rank_clusters_low_to_high(pred_baseline_pd, "cluster_baseline")
pred_baseline_pd["cluster_name"] = pred_baseline_pd["cluster_baseline"].map(cluster_map_kmeans)

print(f"=== MODEL 1: KMeans (Baseline, Unscaled) === Silhouette: {silhouette_baseline:.4f} | WCSS: {wcss_baseline:.4f}")
print("Cluster label mapping:", cluster_map_kmeans)
print(pred_baseline_pd["cluster_name"].value_counts())
```

```python
#DBSCAN
features_pd_full = df_scaled.select("line_code", "week_number", "disruption_count", "features_scaled").toPandas()
X = np.array(features_pd_full["features_scaled"].tolist())

best = None
for eps in np.arange(0.30, 3.01, 0.05):
    labels = DBSCAN(eps=eps, min_samples=5).fit_predict(X)
    non_noise = labels != -1
    n_clusters = len(set(labels[non_noise]))
    if n_clusters == 3 and non_noise.sum() > 3:
        sil = silhouette_score(X[non_noise], labels[non_noise])
        if best is None or sil > best[1]:
            best = (eps, sil, labels)

if best is not None:
    eps_chosen, silhouette_dbscan, dbscan_labels = best
    print(f"DBSCAN: eps={eps_chosen:.2f} chosen — yields exactly 3 clusters, Silhouette={silhouette_dbscan:.4f}")
else:
    print("No eps in range gave exactly 3 clusters — falling back to centroid-merging.")
    dbscan_labels = DBSCAN(eps=1.0, min_samples=5).fit_predict(X)
    non_noise = dbscan_labels != -1
    labels_work = dbscan_labels.copy()
    while len(set(labels_work[labels_work != -1])) > 3:
        ids = sorted(set(labels_work[labels_work != -1]))
        centroids = {c: X[labels_work == c].mean(axis=0) for c in ids}
        pairs = [(a, b, np.linalg.norm(centroids[a] - centroids[b]))
                 for i, a in enumerate(ids) for b in ids[i+1:]]
        a, b, _ = min(pairs, key=lambda p: p[2])
        labels_work[labels_work == b] = a  # merge b into a
    dbscan_labels = labels_work
    non_noise = dbscan_labels != -1
    silhouette_dbscan = silhouette_score(X[non_noise], dbscan_labels[non_noise])

features_pd_full["cluster_dbscan"] = dbscan_labels
features_pd = features_pd_full[features_pd_full["cluster_dbscan"] != -1].copy()  # drop remaining noise points

n_clusters_dbscan = len(set(features_pd["cluster_dbscan"]))
n_noise = (dbscan_labels == -1).sum()

cluster_map_dbscan = rank_clusters_low_to_high(features_pd, "cluster_dbscan")
features_pd["cluster_name"] = features_pd["cluster_dbscan"].map(cluster_map_dbscan)

print(f"\n=== MODEL 2: DBSCAN (Advanced, Density) === Clusters: {n_clusters_dbscan} | Noise excluded: {n_noise} | Silhouette: {silhouette_dbscan:.4f}")
print("Cluster label mapping:", cluster_map_dbscan)
print(features_pd["cluster_name"].value_counts())
```

```python
#GMM
best_gmm = None
for seed in range(30):
    gmm_try = GaussianMixture(featuresCol="features_scaled", predictionCol="cluster_gmm", k=3, seed=seed)
    model_try = gmm_try.fit(df_scaled)
    pred_try = model_try.transform(df_scaled)
    evaluator_gmm = ClusteringEvaluator(featuresCol="features_scaled", predictionCol="cluster_gmm", metricName="silhouette")
    sil_try = evaluator_gmm.evaluate(pred_try)
    if best_gmm is None or sil_try > best_gmm[0]:
        best_gmm = (sil_try, seed, model_try, pred_try)

silhouette_gmm, best_seed, model_gmm, pred_gmm = best_gmm

pred_gmm_pd = pred_gmm.select("line_code", "week_number", "disruption_count", "cluster_gmm").toPandas()
cluster_map_gmm = rank_clusters_low_to_high(pred_gmm_pd, "cluster_gmm")
pred_gmm_pd["cluster_name"] = pred_gmm_pd["cluster_gmm"].map(cluster_map_gmm)

print(f"\n=== MODEL 3: GMM (Advanced, Probabilistic) === best seed={best_seed} | Silhouette: {silhouette_gmm:.4f}")
print("Cluster label mapping:", cluster_map_gmm)
print(pred_gmm_pd["cluster_name"].value_counts())
```

```python
cluster_map = {0: "Moderate Disruptions", 1: "High Disruptions", 2: "Low Disruptions"}
```

```python
pred_baseline_pd = pred_baseline.select("line_code", "week_number", "cluster_baseline").toPandas()
pred_baseline_pd["cluster_name"] = pred_baseline_pd["cluster_baseline"].map(cluster_map)

features_pd["cluster_name"] = features_pd["cluster_dbscan"].map(cluster_map)

pred_gmm_pd = pred_gmm.select("line_code", "week_number", "cluster_gmm").toPandas()
pred_gmm_pd["cluster_name"] = pred_gmm_pd["cluster_gmm"].map(cluster_map)
```

```python
print("\nCluster mapping applied:")
print(pred_baseline_pd["cluster_name"].value_counts())
```

```python
comparison = pd.DataFrame({
    "Model": ["KMeans (Baseline, Unscaled)", "DBSCAN (Advanced, Density)", "GMM (Advanced, Probabilistic)"],
    "Type": ["Partition-based", "Density-based", "Model-based (EM)"],
    "Silhouette": [silhouette_baseline, silhouette_dbscan, silhouette_gmm],
    "N_Clusters": [3, n_clusters_dbscan, 3],
})
```

```python
print("\n=== MODEL COMPARISON ===")
print(comparison)
comparison.to_csv(r"E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\clustering_model_comparison.csv", index=False)
```

```python
def pca_2d(vector_col):
    return PCA(n_components=2, random_state=42).fit_transform(np.array(vector_col.tolist()))
```

```python
pred_baseline_pca = pred_baseline.select("line_code", "week_number", "features_raw", "cluster_baseline").toPandas()
pred_baseline_pca["cluster_name"] = pred_baseline_pca["cluster_baseline"].map(cluster_map_kmeans)
pred_gmm_pca = pred_gmm.select("line_code", "week_number", "features_scaled", "cluster_gmm").toPandas()
pred_gmm_pca["cluster_name"] = pred_gmm_pca["cluster_gmm"].map(cluster_map_gmm)
```

```python
order = ["Low Disruptions", "Moderate Disruptions", "High Disruptions"]
```

```python
bc = pca_2d(pred_baseline_pca["features_raw"])
pred_baseline_pca["pc1"], pred_baseline_pca["pc2"] = bc[:,0], bc[:,1]
dc = pca_2d(features_pd["features_scaled"])
features_pd["pc1"], features_pd["pc2"] = dc[:,0], dc[:,1]
gc = pca_2d(pred_gmm_pca["features_scaled"])
pred_gmm_pca["pc1"], pred_gmm_pca["pc2"] = gc[:,0], gc[:,1]

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

sns.scatterplot(data=pred_baseline_pca, x="pc1", y="pc2", hue="cluster_name", hue_order=order, palette="tab10", ax=axes[0])
axes[0].set_title("KMeans (Baseline, Unscaled)")
axes[0].legend(title="K-Means")

sns.scatterplot(data=features_pd, x="pc1", y="pc2", hue="cluster_name", hue_order=order, palette="tab10", ax=axes[1])
axes[1].set_title("DBSCAN (Advanced, Density)")
axes[1].legend(title="DBSCAN")

sns.scatterplot(data=pred_gmm_pca, x="pc1", y="pc2", hue="cluster_name", hue_order=order, palette="tab10", ax=axes[2])
axes[2].set_title("GMM (Advanced, Probabilistic)")
axes[2].legend(title="GMM")

plt.tight_layout()
plt.savefig("Figures/clustering_pca_comparison.png", dpi=150, bbox_inches="tight")
plt.show()
```
