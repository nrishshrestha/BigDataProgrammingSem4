```python
import os
import xml.etree.ElementTree as ET
import pandas as pd
```

```python
input_dir = "E:\school(softwarica)\Sem-4\BigDataProject\Data\Fares"
output_path = "E:\school(softwarica)\Sem-4\BigDataProject\Data\FinalCSV\clean_fares.csv"
```

```python
data = []

# removing namespaces for parsing
def strip_namespace(tag):
    return tag.split('}', 1)[-1 if '}' in tag else tag]
```

```python
for filename in os.listdir(input_dir):
    if filename.endswith(".xml"):
        file_path = os.path.join(input_dir, filename)
        try:
            tree = ET.parse(file_path)
            root = tree.getroot()

            #Extracting line reference from the root
            line_ref_elem = root.find('.//{http://www.netex.org.uk/netex}LineRef')
            line_id = line_ref_elem.get('ref') if line_ref_elem is not None else "N/A"

            #finding all price elements
            for price_elem in root.findall('.//{http://www.netex.org.uk/netex}DistanceMatrixElementPrice'):
                amount_elem = price_elem.find('{http://www.netex.org.uk/netex}Amount')

                data.append({
                    "LineID": line_id,
                    "Amount": amount_elem.text if amount_elem is not None else "0",
                    "Currency": "GBP",
                    "Source": filename
                })
        except Exception as e:
            print(f"Error processing {filename}: {e}")
```

```python
# saving output
df_fares = pd.DataFrame(data)
df_fares.to_csv(output_path, index = False)
print(f"Extraction complete. Total records: {len(df_fares)}")
print(f"Distinct lines: {df_fares['LineID'].str.split(':').str[-1].nunique()}")
```
